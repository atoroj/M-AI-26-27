<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Pregunta 2.</span>
<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">¿Cuántas palabras totales hay en los comentarios del corpus?</span>

```python
# Incluye aquí el código generado para poder responder a tu pregunta
comentarios = data[data["TIPO DE MENSAJE"] == "COMENTARIO"].copy()
total_palabras = 0
for doc in nlp.pipe(comentarios["CONTENIDO A ANALIZAR"].fillna("").astype(str)):
    total_palabras += sum(1 for token in doc if not token.is_punct and not token.is_space)

print("Numero de palabras totales: ", total_palabras)
```

<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Pregunta 4.</span>
<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Considerando dos grupos de comentarios (odio y no odio) ¿Cuál el número promedio de palabras en los comentarios de cada grupo?</span>

```python
# Incluye aquí el código generado para poder responder a tu pregunta
comentarios = data[data["TIPO DE MENSAJE"] == "COMENTARIO"].copy()

comentarios["GRUPO"] = comentarios["INTENSIDAD"].apply(
    lambda x: "ODIO" if x > 0 else "NO ODIO"
)

comentarios["NUM_PALABRAS"] = [
    sum(1 for token in doc if not token.is_punct and not token.is_space)
    for doc in nlp.pipe(comentarios["CONTENIDO A ANALIZAR"].fillna("").astype(str))
]

media_por_grupo = comentarios.groupby("GRUPO")["NUM_PALABRAS"].mean()

print(media_por_grupo)

```

<b>Incluye aquí, debajo de la línea, la explicación de tu respuesta</b>
<hr>
 Primero filtro los mensajes de tipo COMENTARIO (lo hago en otras celdas)

 Ahora creo una columna nueva, si INTENSIDAD es mayor a 0 lo agrupo como ODIO y si es 0 como NO ODIO

 Una vez creada creo otra columna con el número de palabras por comentario

 Con groupby() de pandas agrupo el numero de palabras con el grupo que pertecene (ODIO Y NO ODIO)

<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Pregunta 5.</span>
<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Considerando dos grupos de comentarios (odio y no odio) ¿Cuál es el número promedio de oraciones en los comentarios de cada grupo?</span>

```python
# Incluye aquí el código generado para poder responder a tu pregunta
comentarios["NUM_ORACIONES"] = [
    sum(1 for sent in doc.sents)
    for doc in nlp.pipe(comentarios["CONTENIDO A ANALIZAR"].fillna("").astype(str))
]

media_oraciones = comentarios.groupby("GRUPO")["NUM_ORACIONES"].mean()
print(media_oraciones)
```

<b>Incluye aquí, debajo de la línea, la explicación de tu respuesta</b>
<hr>
 La función doc.sents de spacy permite identificar las oraciones de un texto.

 Como antes había divido el dataset en 2 grupos odio y no odio, ahora usando la función sents creo una nueva columna que agrupe el numero de oracion por fila.

 Esto lo muestro de nuevo con la función groupby() de pandas y sacando la media por grupo (odio y no odio)

<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Pregunta 6.</span>
<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Considerando dos grupos de comentarios (odio y no odio) ¿Cuál es el porcentaje de comentarios que contienen entidades NER en cada grupo?</span>

```python
# Incluye aquí el código generado para poder responder a tu pregunta
comentarios["TIENE_NER"] = [
    len(doc.ents) > 0
    for doc in nlp.pipe(comentarios["CONTENIDO A ANALIZAR"].fillna("").astype(str))
] 
porcentaje_ner = comentarios.groupby("GRUPO")["TIENE_NER"].mean() * 100
print(porcentaje_ner)
```

<b>Incluye aquí, debajo de la línea, la explicación de tu respuesta</b>
<hr>
 Ahora, creo una columna para guardar si ese comentario tiene al menos un NER.

 Teniendo esto, agrupo por odio y no odio, hago la media y lo multiplico por 100 para extraer el porcentaje

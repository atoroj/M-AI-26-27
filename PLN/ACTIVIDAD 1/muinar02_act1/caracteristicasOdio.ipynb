<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Pregunta 4.</span>
<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Considerando dos grupos de comentarios (odio y no odio) ¿Cuál el número promedio de palabras en los comentarios de cada grupo?</span>

```python
# Incluye aquí el código generado para poder responder a tu pregunta
comentarios = data[data["TIPO DE MENSAJE"] == "COMENTARIO"].copy()

comentarios["GRUPO"] = comentarios["INTENSIDAD"].apply(
    lambda x: "ODIO" if x > 0 else "NO ODIO"
)

comentarios["NUM_PALABRAS"] = [
    sum(1 for token in doc if not token.is_punct and not token.is_space)
    for doc in nlp.pipe(comentarios["CONTENIDO A ANALIZAR"].fillna("").astype(str))
]

media_por_grupo = comentarios.groupby("GRUPO")["NUM_PALABRAS"].mean()

print(media_por_grupo)

```

<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Característica adicional 1.</span>


<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Los comentarios de ODIO presentan una mayor proporción de adjetivos femeninos</span>

```python
documentos = list(
    nlp.pipe(comentarios["CONTENIDO A ANALIZAR"].fillna("").astype(str))
)

# Contamos los adjetivos femeninos de cada comentario
comentarios["ADJETIVOS_FEMENINOS"] = [
    sum(
        1 for token in doc
        if token.pos_ == "ADJ"
        and "Fem" in token.morph.get("Gender")
    )
    for doc in documentos
]

# Contamos el total de adjetivos de cada comentario
comentarios["TOTAL_ADJETIVOS"] = [
    sum(1 for token in doc if token.pos_ == "ADJ")
    for doc in documentos
]

# Calculamos el porcentaje de adjetivos femeninos
# respecto al total de adjetivos de cada comentario
comentarios["PORCENTAJE_ADJETIVO_FEMENINO"] = (
    comentarios["ADJETIVOS_FEMENINOS"]
    / comentarios["TOTAL_ADJETIVOS"].replace(0, float("nan"))
) * 100

# Calculamos la media de porcentajes por grupo
resultado_adj = comentarios.groupby("GRUPO")["PORCENTAJE_ADJETIVO_FEMENINO"].mean()

# Mostramos los resultados
print(resultado_adj.round(2))
```

Se ha propuesto comprobar si los mensajes de odio presentan una mayor proporcion de adjetivos femeninos.

Primero se han contado los adjetivos femeninos de cada comentario
Después se han contado el total de adjetivos de cada comentario
Se ha calculado el procentaje de adjetivos femeninos respecto al total
Y finalmente se ha calculado la media por grupo

Además se ha identificado que no se puede extraer una hipotesis valida para esta caracteristica planteada

<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Característica adicional 2.</span>


<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Los comentarios de ODIO utilizan más verbos en imperativo</span>

```python
# Contamos los verbos en imperativo de cada comentario
comentarios["NUM_IMPERATIVOS"] = [
    sum(
        1 for token in doc
        if token.pos_ == "VERB"
        and "Imp" in token.morph.get("Mood")
    )
    for doc in documentos
]

# Comprobamos si cada comentario contiene al menos un imperativo
comentarios["TIENE_IMPERATIVO"] = (
    comentarios["NUM_IMPERATIVOS"] > 0
)

# Calculamos el porcentaje de comentarios con imperativos
# en cada grupo: ODIO y NO ODIO
porcentaje_imperativos = (
    comentarios.groupby("GRUPO")["TIENE_IMPERATIVO"].mean() * 100
)

# Calculamos el número medio de imperativos por comentario
media_imperativos = (
    comentarios.groupby("GRUPO")["NUM_IMPERATIVOS"].mean()
)

# Mostramos los resultados
print("Porcentaje de comentarios con imperativos:")
print(porcentaje_imperativos.round(2))

print("\nMedia de imperativos por comentario:")
print(media_imperativos.round(2))
```

La idea es comprobar si los mensajes de odio utilizan con mayor ferecuencia expresiones que transmiten órdenes.

En spacy se puede utlizar token.morph.get("Mood") para identificar el modo verbal, con Imp sacamos si es imperativo

<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Característica adicional 3.</span>


<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Los comentarios de tipo ODIO contienen más entidades NER tipo person</span>

```python
# Incluye aquí el código generado para poder responder a tu pregunta
documentos = list(
    nlp.pipe(comentarios["CONTENIDO A ANALIZAR"].fillna("").astype(str))
)

comentarios["NUM_PERSONAS"] = [
    sum(1 for ent in doc.ents if ent.label_ == "PER")
    for doc in documentos
]

# Comprobamos si el comentario contiene al menos una persona
comentarios["TIENE_PERSONA"] = comentarios["NUM_PERSONAS"] > 0

# Calculamos el porcentaje de comentarios con entidades PER
porcentaje_personas = (
    comentarios.groupby("GRUPO")["TIENE_PERSONA"].mean() * 100
)

# Calculamos el número medio de entidades PER por comentario
media_personas = (
    comentarios.groupby("GRUPO")["NUM_PERSONAS"].mean()
)

# Mostramos los resultados
print("Porcentaje de comentarios con personas:")
print(porcentaje_personas.round(2))

print("\nMedia de entidades PER por comentario:")
print(media_personas.round(2))
```

<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Característica adicional 4.</span>


<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Los comentarios de ODIO utilizar con mayor frecuencia formas de expresarse en segunda persona que los comentarios de NO ODIO</span>

```python
# Incluye aquí el código generado para poder responder a tu pregunta
documentos = list(
    nlp.pipe(comentarios["CONTENIDO A ANALIZAR"].fillna("").astype(str))
)

comentarios["VERBOS_SEGUNDA"] = [
    sum(
        1 for token in doc
        if token.pos_ in ["VERB", "AUX"]
        and "2" in token.morph.get("Person")
    )
    for doc in documentos
]

# Comprobamos si cada comentario contiene
# al menos un verbo en segunda persona
comentarios["TIENE_SEGUNDA"] = (
    comentarios["VERBOS_SEGUNDA"] > 0
)

# Calculamos el porcentaje de comentarios que
# contienen verbos en segunda persona dentro de cada grupo
resultado_segunda = (
    comentarios.groupby("GRUPO")["TIENE_SEGUNDA"].mean() * 100
)

# Mostramos los resultados
print(resultado_segunda.round(2))
```

<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">¿Es posible utilizar alguna de las características extraídas en las preguntas anteriores para determinar si un mensaje contiene odio? Justifica tu respuesta con el análisis estadístico que consideres necesario.</span>

```python
# Comparamos las cuatro características propuestas por grupo.
grupos = comentarios.groupby("GRUPO")
resumen = pd.DataFrame({
    "Número de comentarios": grupos.size(),
    "Media de adjetivos femeninos (%)": grupos["PORCENTAJE_ADJETIVO_FEMENINO"].mean(),
    "Comentarios con imperativos (%)": grupos["TIENE_IMPERATIVO"].mean() * 100,
    "Media de imperativos": grupos["NUM_IMPERATIVOS"].mean(),
    "Comentarios con personas PER (%)": grupos["TIENE_PERSONA"].mean() * 100,
    "Media de entidades PER": grupos["NUM_PERSONAS"].mean(),
    "Comentarios con segunda persona (%)": grupos["TIENE_SEGUNDA"].mean() * 100,
})
print("Comparación de las características:")
print(resumen.round(2).T.to_string())
print("\nLa media de porcentajes de adjetivos femeninos excluye comentarios sin adjetivos.")

# Regla sencilla: predecimos ODIO si hay algún verbo en segunda persona.
odio_real = comentarios["GRUPO"].eq("ODIO")
odio_predicho = comentarios["TIENE_SEGUNDA"]
tabla = pd.crosstab(
    comentarios["GRUPO"],
    odio_predicho.map({False: "NO ODIO", True: "ODIO"}),
).reindex(index=["NO ODIO", "ODIO"], columns=["NO ODIO", "ODIO"], fill_value=0)
tabla.index.name = "Grupo real"
tabla.columns.name = "Predicción"
print("\nResultados de la regla basada en segunda persona:")
print(tabla.to_string())

aciertos_odio = int((odio_real & odio_predicho).sum())
falsas_alarmas = int((~odio_real & odio_predicho).sum())
odio_no_detectado = int((odio_real & ~odio_predicho).sum())
precision = aciertos_odio / (aciertos_odio + falsas_alarmas) if (aciertos_odio + falsas_alarmas) else 0.0
cobertura = aciertos_odio / (aciertos_odio + odio_no_detectado) if (aciertos_odio + odio_no_detectado) else 0.0
print(f"\nPrecisión: {precision:.2%} de los comentarios señalados como ODIO lo son.")
print(f"Cobertura (recall): {cobertura:.2%} de los comentarios de ODIO son detectados.")
print(f"Porcentaje de ODIO en el conjunto: {odio_real.mean():.2%}")
```
